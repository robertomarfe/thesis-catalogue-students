# Python runtime check
This generic notebook checks that Python and NumPy execute correctly. It is not a thesis proposal or solution.
Run the single code cell. The expected last line is `RUNTIME_CHECK_PASS`.

```python
import sys
import numpy as np
A = np.array([[3., 1.], [1., 2.]])
b = np.array([9., 8.])
x = np.linalg.solve(A, b)
assert np.allclose(x, [2., 3.])
rng = np.random.default_rng(42)
sample = rng.standard_normal(10000)
assert sample.shape == (10000,) and np.isfinite(sample).all()
print("PYTHON=" + sys.executable)
print("NumPy:", np.__version__)
print("Linear system:", x)
print("RUNTIME_CHECK_PASS")
```
