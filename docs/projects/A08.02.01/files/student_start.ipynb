## Setup exercise: dimensions and positive definiteness

Use a small invented covariance and a row defining a relative view. Verify dimensions and the simplex condition for the reference portfolio; the posterior and optimizer remain to be implemented.

```python
import numpy as np
Sigma = np.array([[.002, .0003], [.0003, .003]])
w0 = np.array([.5, .5])
P = np.array([[1., -1.]])
assert np.all(np.linalg.eigvalsh(Sigma) > 0)
assert np.isclose(w0.sum(), 1.) and np.all(w0 >= 0)
assert P.shape == (1, len(w0))
print('Dimension, SPD and reference-weight checks passed.')
```
