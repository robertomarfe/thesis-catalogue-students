## Setup exercise: units and nonoverlapping windows

The dates below are invented. Confirm 30- and 70-minute windows and the conversion of a decimal rate movement to basis points. No equity responses or regression are supplied.

```python
from datetime import datetime, timedelta, timezone
start = datetime(2000, 1, 1, 14, 0, tzinfo=timezone.utc)
statement = (start, start + timedelta(minutes=30))
conference = (start + timedelta(minutes=45), start + timedelta(minutes=115))
assert (statement[1]-statement[0]).total_seconds() == 1800
assert (conference[1]-conference[0]).total_seconds() == 4200
assert statement[1] <= conference[0]
assert .0025 * 10000 == 25
print('Window and unit conventions passed.')
```
