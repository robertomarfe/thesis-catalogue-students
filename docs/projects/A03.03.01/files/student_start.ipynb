## Setup exercise: transition matrix and one-period discounting

Check the shape of a two-state transition matrix and a deterministic CRRA one-period SDF. This does not solve the recursive wealth equilibrium.

```python
import numpy as np
P = np.array([[.9, .1], [.1, .9]])
assert P.shape == (2, 2) and np.allclose(P.sum(axis=1), 1)
beta, gamma, gross_growth = .98, 2., 1.02
m = beta * gross_growth**(-gamma)
assert np.isclose(m * (1/m), 1.)
print('Transition and one-period pricing identity checks passed.')
```
