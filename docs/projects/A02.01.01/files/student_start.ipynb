## Setup exercise: calendar maturity

```python
import numpy as np
origin = np.arange(6)
end = origin + 1
cutoff = 4
eligible = (origin < cutoff) & (end < cutoff)
assert np.array_equal(origin[eligible], [0, 1, 2])
print('Calendar convention check passed.')
```
